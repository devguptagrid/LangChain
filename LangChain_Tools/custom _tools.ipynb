## 1. Custom tools using @tool decorator

```python
from langchain_core.tools import tool
```

```python
# Step 1 - create a function

def multiply(a,b):
    """Multiply two numbers"""
    return a*b
```

```python
# Step 2 - add type hints

def multiply(a:int,b:int)->int:
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 - add tool decorator

@tool
def multiply(a:int,b:int)->int:
    """Multiply two numbers"""
    return a*b
```

```python
result=multiply.invoke({"a":3,"b":5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
from langchain_core.tools import StructuredTool

multiply = StructuredTool.from_function(multiply)

print(multiply.args_schema.model_json_schema())
```

## 2. Using Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a:int=Field(required=True,description='The first number to multiply')
    b:int=Field(required=True,description='The second number to multiply')
```

```python
def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name='multiply',
    description='multiply two numbers',
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({'a':3,'b':5})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

## 3. Using BaseTool class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a:int=Field(reuquired=True,description='The First number to multiply')
    b:int=Field(required=True,description='The Second number to multiply')
    
```

```python
class MultiplyTool(BaseTool):
    name:str='multiply'
    description:str='Multiply two number'
    args_Schema:Type[BaseModel]=MultiplyInput

    def _run(self,a:int,b:int)->int:
        return a*b
```

```python
multiply_tool=MultiplyTool()
```

```python
result=multiply_tool.invoke({'a':3,'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```
